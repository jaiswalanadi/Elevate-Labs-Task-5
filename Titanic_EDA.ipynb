
# Titanic Dataset - Exploratory Data Analysis (EDA)

## 1. Data Overview

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("train.csv")

# Basic info
df.info()
df.describe(include="all")
df.isnull().sum()
```

## 2. Univariate Analysis

### Survived
```python
sns.countplot(x='Survived', data=df)
plt.title('Survival Count')
plt.show()
```

### Sex
```python
sns.countplot(x='Sex', data=df)
plt.title('Gender Distribution')
plt.show()
```

### Pclass
```python
sns.countplot(x='Pclass', data=df)
plt.title('Passenger Class Distribution')
plt.show()
```

### Age
```python
sns.histplot(df['Age'].dropna(), kde=True, bins=30)
plt.title('Age Distribution')
plt.show()
```

### Fare
```python
sns.histplot(df['Fare'], kde=True, bins=30)
plt.title('Fare Distribution')
plt.show()
```

### Embarked
```python
sns.countplot(x='Embarked', data=df)
plt.title('Port of Embarkation')
plt.show()
```

## 3. Bivariate Analysis

### Survived vs Sex
```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Gender')
plt.show()
```

### Survived vs Pclass
```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Survival by Class')
plt.show()
```

### Age vs Survived
```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age vs Survival')
plt.show()
```

### Fare vs Survived
```python
sns.violinplot(x='Survived', y='Fare', data=df)
plt.title('Fare vs Survival')
plt.show()
```

## 4. Multivariate Analysis

### Correlation Heatmap
```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

### Pairplot
```python
sns.pairplot(df[['Survived', 'Pclass', 'Age', 'Fare']], hue='Survived')
plt.show()
```

## 5. Summary of Findings

- Female passengers had a much higher survival rate than males.
- Passengers in 1st class had higher survival rates.
- Younger passengers and those who paid higher fares were more likely to survive.
- Most passengers embarked from port 'S'.
- Age and fare show variability in relation to survival but not linear.
- Cabin and Ticket fields have many missing or unique values; further processing is needed before use.
